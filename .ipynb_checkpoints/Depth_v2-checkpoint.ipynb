```python
import torch
from transformers import pipeline  # Pipeline :- Used to load a pre-trained model and performance
from PIL import Image   # Image processing liberary used for open,save and resize images
import numpy as nup   #  used for numerical operation
import os   # Built interaction with operating system
import matplotlib.pyplot as plat  # USed to crete graphs,charts and plots

# Creating path to the image folder and selecting system for performing operation
folder_image_path = "D:\\Camera_trap\\data\\raw_data\\eccv_18_all_images_sm"

if torch.cuda.is_available():
    processing_device = "cuda"
else:
    processing_device = "cpu"

print(f"System selected {processing_device}")

# System is selected , now lets create depth estimation pipeline
# This will download an pre-trained model , loads into memory and prepared it for working

if processing_device == "cuda":
    pipeline_device = 0
else:
    pipeline_device = -1

pipeline_esitimation_depth = pipeline(
    task="depth-estimation",
    model="depth-anything/Depth-Anything-V2-Base-hf",
    device=pipeline_device
)

print("All models installed correctly and ready to use ")
```

```python
# The average time for whole data set is rougly 5 hours  :-

import time  # Measure time execution
import json  # Usied to write jsn file and update
from datetime import datetime  # Executing date and time current format


# Lets get all the images from dataset and calculatin of total images:--
total_images_dataset = os.listdir(folder_image_path)
total_images = len(total_images_dataset)
print(f"Total images are {total_images}")

images_not_processed = []
images_processed = 0
images_count_miss = 0

# Json log needs to save differently :--
json_path = "D:\\Camera_trap\\data\\processed_data\\depth_maps\\processing_log.json"

# LEts note start time for process
time_start = time.time()

for i, file in enumerate(total_images_dataset):
    # Getting 3 things image name , remove jpg and create npy path
    id_image = file.replace(".jpg", "")
    result_npy = os.path.join(image_depth_path_npy, f"{id_image}.npy")

    # If already processed then it skipp the image
    if os.path.exists(result_npy):
        images_count_miss += 1
        continue

    # Lets process image using try block for error handling:--
    try:
        location_image = os.path.join(folder_image_path, file)
        img = Image.open(location_image)
        img = img.convert("RGB")

        output = pipeline_esitimation_depth(img)
        img_depth_map = output["depth"]
        img_depth_array = nup.array(img_depth_map)

        # Saving.npy and .png differently:--
        nup.save(result_npy, img_depth_array)
        img_depth_map.save(os.path.join(image_depth_path_png, f"{id_image}.png"))

        images_processed += 1

    # Lets see how many images get failed to exexute
    except Exception as e:
        images_not_processed.append(
            {
                "image_failed_id": id_image,
                "error_result": str(e)
            }
        )
        print(f"Failed images = {id_image} - {e}")
        continue

    # This will give us total images which are not processed

    # check process after 500 images :-
    if (i + 1) % 500 == 0:
        # When image count reaches 500 and gives remained 0 then it will execute this
        time_taken = time.time() - time_start

        # Lets find out rate of image provessing:--(Average time taken per image)
        if images_processed > 0:
            rate_images = time_taken / images_processed
        else:
            rate_images = time_taken / 1

        # Total remaining images :--
        images_total_remaining = total_images - (i + 1)

        # Lets get what total time remaining until all images processed :-
        remaining_time_estimated = (rate_images * images_total_remaining) / 60  # Gives us left minutes

        right_now_time = datetime.now().strftime("%H:%M:%S")
        print("Current_time", right_now_time)
        print("Images checked till ", i + 1, "/", total_images)
        print("Images processed", images_processed)
        print("Images skipped", images_count_miss)
        print("Images not processed", len(images_not_processed))
        print("Time remaining", round(remaining_time_estimated, 1), "Minutes")

        # Saving an flag here log if it gets crashed:-
        with open(json_path, "w") as f:
            # Dump method used to write data in json file
            json.dump(
                {
                    "Image processed": images_processed,
                    "Image skipped": images_count_miss,
                    "Images failed": images_not_processed,
                    "Index last": i
                },
                f,
                indent=2
            )

time_end = time.time()  # Noting total time taken

print("===Done===")
print(f"Total time taken for whole process {(time_end - time_start) / 3600:.2f} hours")
print(f"Processed images {images_processed}")
print(f"Images missed {images_count_miss}")
print(f"Images not processed {images_not_processed}")


# Saving the final log:00
with open(json_path, "w") as f:
    json.dump(
        {
            "Image processed": images_processed,
            "Image skipped": images_count_miss,
            "Images failed": images_not_processed,
            "Last index": total_images - 1
        },
        f,
        indent=2
    )

# Saving log for failed images :-
if images_not_processed:
    print(f"Total images which are not processed {len(images_not_processed)}")
```

```python
# We have got our depth map for whole dataset 
# Next thing before implementing depth fusion model is that how do we need to implement our model :--
# we need to build an separate dataset class for fusiuon model bcz orignal dataset used by ResBet 50 have RGB channels 
# Now we dont have an specific logic to load depths maps . 
# For my pubshiable result which is (RGB Alone vs RGB+ Depth ) the question is can adding depth model imporves camera trap data anaylysis or not 
# Hence we need both model to be perform at same time for better comparision 
# SO we need to create our separate database exactly like orignal one 
# Its an sequence dataset just like orignalm dataset 
```

```python
from torchvision import transforms 


#LEts define mean value used for normalization 
# First we need to re-normalized each image because initially we have converted image into single 4-channel 
# tensor. channels contain raw pixel values ij range of 0 to 1 
# Hence we will perform normalization using pre-defined values of mean and standard deviation
# Hence we dont want to perform with same input values imags which we performeed in resnet 50
```

```python
defined_mean = [0.485, 0.456, 0.406, 0.5]
defined_deviation =  [0.229, 0.224, 0.225, 0.225]
# Pipeline for transformation
# combining one or more transformations into single pipeline 
re_transform = transforms.Compose([
    transforms.Normalize(mean = defined_mean , std = defined_deviation)
    # mean,std are values of old transformation
])
print("Done with normalization")
```

```python
import torch
from torch.utils.data import Dataset, Subset, DataLoader

# Extracting three things from torch.utils.data
# 1. Dataset :-- This is used to create an duplicate dataset
# 2. Subset :- This is used to create an small portion of dataset :--
# 3. Dataloaders :-- Dataloaders are used to create an batch of images in an random number ,
# Such as images of batch 32

import numpy as np
from PIL import Image  # Used to open, resize and process image
import os  # To get work with operating system
import json

# Lets create an class for dataset of fusion model :--
class Dataset_fusion(Dataset):

    def __init__(self, file_image, npy_file, annotations_image_path, transform=None):

        self.file_image_path = file_image
        self.npy_file = npy_file
        self.transform = transform

        with open(annotations_image_path, "r", encoding="utf-8") as f:
            self.annotations = json.load(f)

        # Lets
        id_category = []

        for annotation in self.annotations:
            id_category.append(annotation["category_id"])

        labels_unique = set(id_category)

        # Ids after sorted :--
        ids_done = sorted(labels_unique)

        # We need to sort our ids in an proper numerical format
        # Hence needs t create new dictonary
        self.label_map = {}

        count = 0

        for id in ids_done:
            self.label_map[id] = count
            count = count + 1

    # Returns total annotations which is required for data_loaders to count images
    def __len__(self):
        return len(self.annotations)
        length = len(self.annotations)
        print(f"length {length}")
    
    
    # Now lets get one sample image and implememnt rgb + depth map with one label
    def __getitem__(self, index):

        # Selecting one particular image
        values_annotations = self.annotations[index]

        id_test_image = values_annotations["image_id"]

        # Gets that image for that annotation through image_id
        # Creating the path of that image and openning it using Image
        rgb_only_path = os.path.join(self.file_image_path, id_test_image + ".jpg")

        rgb_only_image = Image.open(rgb_only_path).convert("RGB")

        # We need to resize image the image for model input :--
        rgb_only_image = rgb_only_image.resize((224, 224))

        # Now we need to convert this image into an numerical value can be added to model as input
        rgb_image_num = np.array(rgb_only_image).astype(np.float32) / 255.0

        # This will convert image into an NumPy array
        # This will convert datatype to 32_bit floating point number for deep model learning

        rgb_depth_path = os.path.join(self.npy_file, id_test_image + ".npy")

        rgb_depth_num = np.load(rgb_depth_path).astype(np.float32)

        # This will read file in an NumPy array and stores it into rgb_depth_num
        # Now we got the depth map shape and image in pixels (224*224)

        depth_img = Image.fromarray(rgb_depth_num.astype(np.uint8))
        depth_img = depth_img.resize((224, 224))

        rgb_depth_num = np.array(depth_img).astype(np.float32) / 255.0

        # To combine this with an RGB image we need an channel dimension :-
        rgb_depth_num = np.expand_dims(rgb_depth_num, axis=-1)

        # Now the Depth map got that one channel added 224*224*1
        # RGB already have 224*224*3
        # Hence now we got the 4 channels are RGB + Depth
        # And we got our RGDB image which is RGB + Depth image

        fused = np.concatenate([rgb_image_num, rgb_depth_num], axis=-1)

        # Convert the NumPy array into pytorch tensor
        # As model only accepts pytorch image before training :--
        RGDB_img_fused = torch.from_numpy(fused)

        # Tensor shape -> height * width * channels
        # New input -> channels * height * width
        RGDB_img_fused = RGDB_img_fused.permute(2, 0, 1)

        RGDB_img_fused = RGDB_img_fused.float()

        if self.transform:
            RGDB_img_fused = self.transform(RGDB_img_fused)

        label = self.label_map[values_annotations["category_id"]]

        return RGDB_img_fused, label
```

```python
from torchvision import transforms
standard_deviation = [0.229, 0.224, 0.225]
Mean = [0.485, 0.456, 0.406]

transform_image = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(Mean, standard_deviation)
])






from torch.utils.data import Dataset
import json
import os
from PIL import Image

class Camera_Dataset(Dataset):
    def __init__(self, images_folder, path_annotations, transform=None):
        self.images_folder = images_folder
        self.path_annotations = path_annotations
        self.transform = transform
        with open(path_annotations, "r", encoding="utf-8") as f:
            self.annotations = json.load(f)
        final_ids = sorted(set(a["category_id"] for a in self.annotations))
        self.label_map = {cat_id: slot for slot, cat_id in enumerate(final_ids)}

    def __len__(self):
        return len(self.annotations)

    def __getitem__(self, index):
        annotation_data = self.annotations[index]
        test_image_path = os.path.join(self.images_folder, annotation_data["image_id"] + ".jpg")
        Img = Image.open(test_image_path)
        Img = Img.convert("RGB")
        lb = self.label_map[annotation_data["category_id"]]
        if self.transform:
            Img = self.transform(Img)
        return Img, lb

d = Camera_Dataset(
    r"D:\Camera_trap\data\processed_data",
    r"D:\Camera_trap\data\processed_data\prepared_annotations.json",
    transform=transform_image
)
print(f"d size = {len(d)}")
```

```python
# Now we have created an dataset class and now we will execute the old train/test split
# Which we have created earlier for our ResNet 50 model

import torch
torch.manual_seed(42)
from torch.utils.data import random_split
dataset_train_size = int(len(d) * 0.8)
dataset_test_size = len(d) - dataset_train_size
dataset_train, dataset_test = random_split(d, [dataset_train_size, dataset_test_size])
print(f"Train size = {len(dataset_train)}")
print(f"Test size = {len(dataset_test)}")
# Now we have created an dataset class and now we will execute the old train/test split
# Which we have created earlier for our ResNet 50 model
dataset_fusion = Dataset_fusion(
    r"D:\Camera_trap\data\processed_data",
    r"D:\Camera_trap\data\processed_data\depth_maps\npy",  # Folder containing all depth maps
    r"D:\Camera_trap\data\processed_data\prepared_annotations.json",  # Annotation file loaded
    transform=re_transform 
)
print(f"Dataset size fusion {len(dataset_fusion)}")
# This will give us exact number of images present inside new dataset
model_train = Subset(dataset_fusion, dataset_train.indices)  # Only training images will be stored
model_test = Subset(dataset_fusion, dataset_test.indices)    # Only testing images will be stored
# Hence we got the dataset for both lets load it into data_loaders :--
data_loader_train = DataLoader(model_train, batch_size=32, shuffle=True)
data_loader_test = DataLoader(model_test, batch_size=32, shuffle=False)
# Lets go through each image
iterator = iter(data_loader_train)
images, labels = next(iterator)
print("Image shape:", images.shape)
print("Labels shape:", labels.shape)
```

```python
import torchvision.models as models  # Importing pre trained ResNet 50
import torch
import torch.nn as nn  # Import neural network use for layers

# Model function :-
def model_fusion(num):

    model = models.resnet50(weights="IMAGENET1K_V1")  # Load pre-trained model and weighs from ImageNet

    # Convulation layer :-- This layer provides an filter to input images to detect patterns like edges and shapes.
    previous_conv = model.conv1

    # Creating a new convulation layer because old layeer only have old 3 input channels
    # Now new one will accept 4 input channels
    fusion_conv = nn.Conv2d(
        4,  # Input channels
        previous_conv.out_channels,  # Output channels will remain same
        kernel_size=previous_conv.kernel_size,  # Kernal size will remain same
        stride=previous_conv.stride,
        padding=previous_conv.padding,
        bias=False
    )

    with torch.no_grad():
        fusion_conv.weight[:, :3, :, :] = previous_conv.weight  # Getting pre-trained RGB weights

        # Initillizing the weights of orignal convulation layer to new one
        fusion_conv.weight[:, 3:, :, :] = previous_conv.weight[:, :1, :, :]

    model.conv1 = fusion_conv

    model.fc = nn.Linear(model.fc.in_features, num)

    return model


model = model_fusion(num=13).to(processing_device)

print(" === All done === ")
print("Model is ready to use")
```

```python
# Training phasse of model :--
import torch.optim as optim # Importing optimizer used to update module weights 
loss_function = nn.CrossEntropyLoss()
# This function calculates loss (Calculated the wrong predicted)
opti = optim.Adam(model.parameters(), lr = 0.0001)
# Uses the adam optimizer with learning rate (Learning-rate :- This will control how much weight should ne changed)
round = 5
# 5 Times a model should train for accurate learning 
for num in range(round):
    # LEts get modl into training mode :--
    model.train()
    current_loss = 0.00
    for images, labels in data_loader_train:
        images = images.to(processing_device)
        labels = labels.to(processing_device)
        opti.zero_grad() # Claers the gradient so it must get fresh before every size
        # Forward pass :--
        results = model(images)
        loss = loss_function(results, labels) # CAlculate the loss after training
        loss.backward()
        opti.step() # Update the model weiht after calculating loss 
        current_loss = current_loss + loss.item()
        # This will give us total loss at the end in number format 
    current_round = num + 1
    total_rounds = round
    size_batch = len(data_loader_train)
    average_loss = current_loss / size_batch
    print(f"Round [{current_round}/{total_rounds}]")
    print(f"Average calculated loss : {average_loss:.4f}")
    print("Model training completed")
    # Saving model :--
torch.save(model.state_dict(), r"D:\Camera_trap\models\fusion_resnet50_trained.pth")
print("Model saved")
```

```python
# Training completed now we need to evulate 
from sklearn.metrics import confusion_matrix , classification_report # used to get Confusion_matrix 
import matplotlib.pyplot as plt # Display graphs annd figures 
import seaborn as sns # Creates heatmap and confusion matrix

# Putting evulation :--
model.eval() # Lets put model in evulate model 

model_actual_total_labels = []
model_prediction_label = []

with torch.no_grad():
# Gradient computation is tyruned off becanuse model is in evulation mode 
# Saves memory and speeds up prediction 
    # Going through dataset
    for images, labels in data_loader_test:
        # Moves both images to processing device which is GPU 
        images , labels = images.to(processing_device), labels.to(processing_device)

        # Foward pass:- means  images are passing throguh Fusion ResNet 50 model
        output_scores = model(images)

        # The output scores for every class will be stored in this vatiable 
        # Highest class will be selected 
        _, _predicted = torch.max(output_scores ,1)

        # torch.max select max score (_ ignores all other value only class value is needed)

        # Store actuall labels to match with the predicted one 
        model_actual_total_labels.extend(labels.cpu().numpy())

        # This convert predicted labels into GPU_TENSOR-> CPU_TENSOR-> NUMPY ARRAY-> FINAL prediction list 
        model_prediction_label.extend(_predicted.cpu().numpy())

        #The labels are compared against predicted labels 

# Lets cretae confusion matrix for this :--
evulation_matrix = confusion_matrix(
    model_actual_total_labels,
    model_prediction_label
)

# Matrix contains :-- 
# ROWS :-- True Classes 
# Coloums :-- Predicted Classes
total_class_fusion = [
    str(name)
    for name , index in sorted(
        d.label_map.items(),
        key = lambda x:x[1]
    )
]

# Create desing of confusion matrix
plt.figure(figsize = (10,8))

# Heatmap:--
sns.heatmap(
    evulation_matrix,
    annot = True,
    fmt = "d",
    cmap = "Blues",
    xticklabels = total_class_fusion,
    yticklabels = total_class_fusion
)

plt.xlabel("Model output")
plt.ylabel("Actuall value")
plt.title("Evulation matrix - Fusion Model(RGB + DEPTH)")
plt.tight_layout()
plt.show()

# Generate classification report :--
print(classification_report(
    model_actual_total_labels,
    model_prediction_label,
    labels = list(range(len(total_class_fusion))),
    target_names = total_class_fusion
))
```
