```python
# Exploring data-set
#Liberary used is os to read data 

import os 

# Path of processed data after extraction (Changing path of extracte2d data)
data_processed_path = "D:\\Camera_trap\\data\\processed_data"

count = 0

# Counting total numbers of images by using os.walk function  :--
for root,dirs,files in os.walk(data_processed_path):
    for file in files:
        if file.endswith("JPG"):
            count = count + 1

# Printing total images = 860
print(f" Total images = {count}")
        
```

```python
# Checking annotation on file:--
import json
annotation_data_path = "D:\\Camera_trap\\data\\annotations\\_annotations.coco.json"
# opening of json file:--
with open(annotation_data_path , 'r') as f:
    data_json = json.load(f)

# Printing keys in annotation file:--

# .key() methodn in dictonary is used to fetch keys in json file
for key in data_json.keys():
    print(f" Keys in json file {key}")

# Counting each item inside each key
for key in data_json.keys():
    print(f"key inside {key} holds Value {len(data_json[key])}")


# Len is used to fetch items present inside each key

    
    
    
```

```python
# Identifying animal catogories:--

for category in data_json['categories']:
    print(f"ID: {category['id']}  Name: {category['name']}")
# Used catogories inside json folder to fetch total types of animals:--

```

```python
# Now we need to check how many images each animals has:--
# To check weather the model will be biased or not
from collections import Counter
total_category_count = Counter()
for annotation in data_json['annotations']:
    total_category_count[annotation['category_id']] += 1

for category in data_json['categories']:
    count = total_category_count[category['id']]
    print(f"  {category['name']}: {count}")
```

```python
# Now it can be noticed thatb there are very low number of female-lions,hyena,hooney-badger,jackal,leopard and wild dog prsent 
# Hence this can be an class imblance:--
# So we will go with 4 well represented classes 
# Those are :--
# 1. Elephant-161
#2. Monkey - 108
# 3. Impala-35
# 4. Waterbuck -11
```

```python
logical_categories =[]
# We will store all valid catogories in this:--

for category in data_json['categories']:
    num = total_category_count[category['id']]
    if num > 10:
        logical_categories.append(category)
        print(f"We will keep :{category['name']}  ({num}annotations)")
    else:
        print(f"We are removing: {category['name']} ({num}annotations)")

        
        
# Hence the animals whose presence is more than 10 will be counted or else will negleted:--
```

```python
# Lets imagine some animals with boundinf boxes:--
import cv2 # Used for reading and processing images:--
import matplotlib.pyplot as plt # This is used to create an graphical visulation of data :--

# Creating an bounding box on image for testing:--
test_image_bound=data_json['images'][0]
print(test_image_bound)

# Building full path to that image:--
test_image_bound_path = os.path.join("D:\\Camera_trap\\data\\processed_data\\valid", test_image_bound['file_name'])
print(test_image_bound_path)
```

```python
# TO draw to bbox we need to find the annotation of that image:--
# To find the exact annotation
annotation_matching_image = []
for annotation in data_json['annotations']:
    if annotation["image_id"] == test_image_bound["id"]:
        annotation_matching_image.append(annotation)
# Now we have matched the image annotation with the image using id:--
print(f"annotaion for the image are {len(annotation_matching_image)}")
```

```python
# Now we got the annotaion we need to convert that into our input parameters for bbbox:--

parameters_bbbox = annotation_matching_image[0]["bbox"]

a, b, w, h = parameters_bbbox

# Writing down the parameters:--
a1 = int(a)
b1 = int(b)
a2 = int(a + w)
b2 = int(b + h)

print(a1, b1, a2, b2)
```

```python
# Now we will draw rectangle using cv2.rectangle method:--

# Loading image :--
img= cv2.imread(test_image_bound_path)
img =cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert image into an RGB 

cv2.rectangle(img, (a1, b1), (a2, b2), (255, 0, 0), 3)  # Only red colour

# Display image :--
plt.figure(figsize=(10, 6))  
plt.imshow(img)
plt.axis("off")
plt.title(test_image_bound["file_name"])
plt.show()

```

```python
def images_with_boxes(image_id):
    # This will fetch the image through dataset:--
    info_img = ""
    for image in data_json["images"]:
        if image["id"] == image_id:
            info_img = image
            
    # This will get the annotation of that image:--
    # This will store the annotation of image 
    info_annotation = []
    for annotation in data_json["annotations"]:
        if annotation["image_id"] == image_id:
            info_annotation.append(annotation)
    # Create image path :--
    info_image_path = os.path.join("D:\\Camera_trap\\data\\processed_data\\valid", info_img["file_name"])

    # Read and change color 
    info_image_display = cv2.imread(info_image_path)
    info_image_display = cv2.cvtColor(info_image_display, cv2.COLOR_BGR2RGB)
    
    # Bounding boxes on image:--
    for annotation in info_annotation:
        bbox = annotation["bbox"]
        x, y, w, h = bbox
        x1 = int(x)
        y1 = int(y)
        x2 = int(x + w)
        y2 = int(y + h)
        cv2.rectangle(info_image_display, (x1, y1), (x2, y2), (255, 0, 0), 3)

    plt.figure(figsize=(10, 6))
    plt.imshow(info_image_display)
    plt.axis("off")
    plt.title(info_img["file_name"])
    plt.show()
```

```python
# Calling the function:--
images_with_boxes(44)

# Now we can clearly see the red rectangle boxes around the animal in every images
# We can acess any image with red box using its image_id
```
