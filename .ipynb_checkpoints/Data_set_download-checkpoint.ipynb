```python
# Extraction of data-set from zip and saving it inside orignal extraction path

import zipfile
import os

# Creating path for storing dataset:--
raw_path_dataset = "D:\\Camera_trap\\data\\raw_data\\Camera Trap Data v1.coco.zip"
extract_path_dataset = "D:\\Camera_trap\\data\\processed_data"

# Extract process:--
with zipfile.ZipFile(raw_path_dataset , 'r') as zip_ref:
    zip_ref.extractall(extract_path_dataset)
    print("Extraction completed")


# Inside the dataset:--
for root,dirs,files in os.walk(extract_path_dataset):
    level = root.replace(extract_path_dataset, '').count(os.sep)
    indent = '' *2 * level
    print(f"{indent} ,{ os.path.basename(root)}/")
    subindent = ' ' *2 * (level +1)
    for file in files[:3]:
        size = os.path.getsize(os.path.join(root, file))
        print(f"{subindent} ,and {len(files)-3} more files")
        if len(files) >3:
            print(f"{subindent}  ,and {len (files)-3} more files") 
            
```
